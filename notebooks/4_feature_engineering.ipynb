# Feature Engineering

## Objective

The objective of this notebook is to prepare application-ready datasets from the cleaned Indian Technology Jobs dataset.

Instead of generating insights, this notebook transforms the cleaned data into structures that can be directly consumed by the CareerLens AI application.

---

## Tasks

- Load the cleaned dataset
- Engineer application-ready datasets
- Create datasets for market insights
- Create datasets for career recommendations
- Create datasets for salary analysis
- Export processed datasets

---

## Expected Output

Multiple processed datasets optimized for the CareerLens AI Streamlit application.

```python
import pandas as pd

pd.set_option("display.max_columns", None)

df = pd.read_csv(
    "../data/processed/careerlens_cleaned.csv",
    keep_default_na=False
)

df.head()
```

```python
dashboard_kpis = pd.DataFrame({

    "Metric":[

        "Total Jobs",

        "Companies",

        "Cities",

        "Job Roles",

        "Skill Domains"

    ],

    "Value":[

        len(df),

        df["company_name"].nunique(),

        df["primary_city"].nunique(),

        df["job_title"].nunique(),

        df["skill_domain"].nunique()

    ]

})

dashboard_kpis
```

```python
dashboard_kpis.to_csv(
    "../data/processed/dashboard_kpis.csv",
    index=False
)
```

```python
role_summary = (

    df.groupby("job_title")

      .agg(

          Number_of_Jobs=("job_title","count"),

          Average_Rating=("company_rating","mean"),

          Average_Salary=("salary_midpoint_lpa","mean")

      )

      .round(2)

      .sort_values(

          "Number_of_Jobs",

          ascending=False

      )

      .reset_index()

)

role_summary.head()
```

```python
role_summary.to_csv(
    "../data/processed/role_summary.csv",
    index=False
)
```

```python
city_summary = (

    df.groupby("primary_city")

      .agg(

          Number_of_Jobs=("primary_city","count"),

          Average_Rating=("company_rating","mean"),

          Average_Salary=("salary_midpoint_lpa","mean")

      )

      .round(2)

      .sort_values(

          "Number_of_Jobs",

          ascending=False

      )

      .reset_index()

)

city_summary.head()
```

```python
city_summary.to_csv(
    "../data/processed/city_summary.csv",
    index=False
)
```

```python
company_summary = (

    df.groupby("company_name")

      .agg(

          Number_of_Jobs=("company_name","count"),

          Average_Rating=("company_rating","mean")

      )

      .round(2)

      .sort_values(

          "Number_of_Jobs",

          ascending=False

      )

      .reset_index()

)

company_summary.head()
```

```python
company_summary.to_csv(
    "../data/processed/company_summary.csv",
    index=False
)
```

### Skill counts

Skill cleaning (casing, junk removal, generic-term filtering) happens once, in `2_data_cleaning.ipynb`. That notebook is the single source of truth for skill counts, so this notebook just loads its output instead of re-deriving it.

```python
skill_summary = pd.read_csv("../data/processed/clean_skill_frequency.csv")

skill_summary.to_csv(
    "../data/processed/skill_summary.csv",
    index=False
)

skill_summary.head(20)
```

```python
KEEP_SKILLS = {
    # Programming
    "Python", "Java", "C++", "C", "C#", "JavaScript", "TypeScript",
    "R", "Scala", "Go", "Ruby", "PHP", "Kotlin", "Swift",

    # Databases
    "SQL", "MySQL", "PostgreSQL", "MongoDB", "Oracle",
    "SQL Server", "Redis", "Cassandra", "Snowflake", "BigQuery",

    # Data / ML / AI
    "Machine Learning", "Deep Learning", "Natural Language Processing",
    "Computer Vision", "Generative AI", "Artificial Intelligence",
    "LLM", "RAG", "MLOps", "Data Mining", "Statistical Modeling",
    "Statistics", "PyTorch", "TensorFlow", "Scikit-learn",
    "NumPy", "Pandas", "Keras",

    # Data Engineering
    "Data Engineering", "ETL", "Spark", "PySpark", "Hadoop",
    "Kafka", "Airflow", "Databricks", "Hive", "Data Warehousing",
    "Azure Data Factory",

    # Cloud
    "AWS", "Microsoft Azure", "Azure", "GCP", "Google Cloud",
    "Cloud Computing",

    # DevOps / Infrastructure
    "Docker", "Kubernetes", "Linux", "Git", "GitHub",
    "CI/CD", "Jenkins", "Terraform", "Ansible",

    # Web / Backend
    "React", "Angular", "Vue", "Node.js", "Django", "Flask",
    "FastAPI", "Spring Boot", "REST API", "APIs",
    "HTML", "CSS",

    # Visualization / BI
    "Power BI", "Tableau", "Excel",

    # Enterprise / other technical platforms
    "SAP", "Salesforce",

    # Technical concepts
    "Data Structures", "Algorithms", "System Design",
    "Database Design", "Microservices", "Software Testing",
    "Unit Testing", "SDLC", "Version Control", "Big Data",
    "SAS", "Selenium"
}
```

```python
# Case-insensitive match: skill_summary's casing comes from whichever raw spelling
# was most common in the scrape (e.g. "power bi"), which won't always match how
# KEEP_SKILLS is typed. Match on lowercase, but display using KEEP_SKILLS's casing.
KEEP_SKILLS_LOWER = {s.lower(): s for s in KEEP_SKILLS}

advisor_skills = skill_summary[
    skill_summary["Skill"].str.lower().isin(KEEP_SKILLS_LOWER)
].copy()
advisor_skills["Skill"] = advisor_skills["Skill"].str.lower().map(KEEP_SKILLS_LOWER)

# collapse any remaining case/duplicate rows (e.g. "power bi" + a stray "Power Bi")
advisor_skills = (
    advisor_skills.groupby("Skill", as_index=False)["Frequency"].sum()
    .sort_values("Frequency", ascending=False)
    .reset_index(drop=True)
)

advisor_skills.to_csv(
    "../data/processed/advisor_skills.csv",
    index=False
)

advisor_skills
```
