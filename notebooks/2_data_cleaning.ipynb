### Normalizing skill spellings

`skills_required` is free text, so the same skill shows up under many spellings — `.Net` / `.NET` / `asp.net` / `ASP.Net`. A handful of raw variants also include stray numbers like `"10635"` and full sentence fragments like `"Business Analyst in IT / digital transformation projects"`.

The pass below does a systematic clean in one place: drop junk tokens, merge case/punctuation variants of the same skill into one canonical label, drop generic non-skill terms and job-title fragments, and drop skills so rare (appearing in only 1-2 of 23,000+ postings) they aren't useful for matching anyway.

This is the **only** place skill cleaning happens in the pipeline — later notebooks reuse `clean_skill_frequency.csv` produced here instead of re-deriving skill counts from scratch.

```python
import re
from collections import Counter, defaultdict

def split_raw(skills_str):
    if pd.isna(skills_str) or skills_str in ("", "Not Available"):
        return []
    return [s.strip() for s in skills_str.split(",") if s.strip() != ""]

raw_token_freq = Counter()
for skills in df["skills_required"]:
    raw_token_freq.update(split_raw(skills))

def is_junk(tok):
    # pure numbers/symbols, e.g. "10635", "277"
    if re.fullmatch(r"[\d.,+%\-]+", tok):
        return True
    # sentence-fragment style entries, not a discrete skill
    if len(tok.split()) > 6:
        return True
    # stray leading-punctuation scrape artifacts
    if tok.startswith(". "):
        return True
    # single stray characters, except real 1-letter language names
    if len(tok) == 1 and tok.upper() not in {"R", "C"}:
        return True
    return False

kept_tokens = {tok for tok in raw_token_freq if not is_junk(tok)}

# Fold case/punctuation variants of the same skill (".Net" / ".NET" / "asp.net") into one canonical label,
# choosing the most frequently occurring original spelling as the display form.
def fold_key(tok):
    return re.sub(r"[^a-z0-9]", "", tok.lower())

groups = defaultdict(list)
for tok in kept_tokens:
    groups[fold_key(tok)].append(tok)

canonical_for = {}
for key, variants in groups.items():
    if key == "":
        continue
    best = max(variants, key=lambda v: raw_token_freq[v])
    for v in variants:
        canonical_for[v] = best

print(f"Raw unique tokens: {len(raw_token_freq)}")
print(f"Kept after junk filter: {len(kept_tokens)}")
print(f"Merged into canonical skills: {len(set(canonical_for.values()))}")
```

```python
# Generic concepts / disciplines that aren't a specific tool or technology.
# Matched case-insensitively against the canonical label (see fix below) so casing
# differences in the source data can no longer cause silent misses.
GENERIC_TERMS = {
    "Analytical", "Analysis", "Analytics", "Business Analysis",
    "Project Management", "Data", "Data Management", "Data Processing",
    "Data Modeling", "Data Quality", "Coding", "Programming", "Development",
    "Management", "Consulting", "Cloud", "Architecture", "Operations",
    "Workflow", "Research", "Sales", "Training", "Compliance", "Forecasting",
    "Process Improvement", "Risk Management", "Healthcare", "Computer Science",
    "Business Development", "Customer Service", "Market Research", "Supply Chain",
    "Information Technology", "Lead Generation", "Basic",
}
GENERIC_TERMS_LOWER = {t.lower() for t in GENERIC_TERMS}

# Job-title filter: any canonical skill that IS or ENDS WITH a role word is a title,
# not a technology ("Data Scientist" -> drop; "Data" alone already covered above;
# "Python", "AWS", "Power BI" don't match any of these words, so they're untouched).
# Using \b word boundaries so this won't match inside unrelated words
# (e.g. "Architect" doesn't match "Architecture").
import re

ROLE_WORDS = [
    "Analyst", "Scientist", "Engineer", "Developer", "Manager", "Consultant",
    "Architect", "Administrator", "Specialist", "Executive", "Coordinator",
    "Officer", "Director", "Lead", "Head", "Trainer", "Intern", "Associate",
    "Representative", "Supervisor",
]
ROLE_TITLE_PATTERN = re.compile(r"\b(" + "|".join(ROLE_WORDS) + r")\b", re.IGNORECASE)

# Safety-valve allowlist: canonical skills that would otherwise match ROLE_TITLE_PATTERN
# but are genuine named tools/products, not job titles. Empty today; add entries here
# if a real technology name is ever caught by the pattern.
ROLE_PATTERN_ALLOWLIST = set()

def is_generic_or_title(canon):
    if canon.lower() in GENERIC_TERMS_LOWER:
        return True
    if canon in ROLE_PATTERN_ALLOWLIST:
        return False
    if ROLE_TITLE_PATTERN.search(canon):
        return True
    return False

flagged = {c for c in set(canonical_for.values()) if is_generic_or_title(c)}
print(f"Canonical skills flagged as generic/job-title: {len(flagged)} of {len(set(canonical_for.values()))}")
```

```python
# Verify against the exact terms that silently failed before the fix
for skill in [
    "Analytics", "Analytical", "Business Analyst", "Computer Science",
    "Coding", "Data", "Business Analysis", "Project Management",
    "Data Analyst", "Data Scientist",
]:
    canon = canonical_for.get(skill, skill)
    print(f"{skill!r:22s} -> canonical {canon!r:22s} | removed now? {is_generic_or_title(canon)}")

print()
# Spot-check that real technologies are NOT touched
for skill in ["Python", "SQL", "AWS", "Power BI", "Spark", "TensorFlow", "Tableau", "Java"]:
    canon = canonical_for.get(skill, skill)
    print(f"{skill!r:22s} -> canonical {canon!r:22s} | removed now? {is_generic_or_title(canon)}")
```

```python
def clean_skill_list(skills_str):
    out, seen = [], set()
    for tok in split_raw(skills_str):
        canon = canonical_for.get(tok)
        if canon and not is_generic_or_title(canon) and canon not in seen:
            seen.add(canon)
            out.append(canon)
    return out

cleaned_lists = df["skills_required"].apply(clean_skill_list)

# Drop skills so rare across the corpus (<3 postings out of 23,000+) that they add
# noise to the skill dropdown/summary without being useful for matching.
MIN_FREQ = 3
canonical_freq = Counter(s for lst in cleaned_lists for s in lst)
keep_final = {s for s, freq in canonical_freq.items() if freq >= MIN_FREQ}

df["skills_required"] = cleaned_lists.apply(
    lambda lst: ", ".join(s for s in lst if s in keep_final)
)
df["skills_count"] = df["skills_required"].apply(lambda s: 0 if s == "" else len(s.split(", ")))

print(f"Final unique skills: {len(keep_final)}")
print(f"Jobs left with zero skills after cleaning: {(df['skills_required']=='').sum()} / {len(df)}")
df["skills_required"].head()
```
