# Data Understanding

## Objective

The objective of this notebook is to explore and understand the raw Indian Technology Jobs dataset before performing any preprocessing.

This notebook focuses on understanding the dataset's structure, quality, and characteristics. No modifications are made to the original data.

---

## Tasks

- Load the dataset
- Inspect its structure
- Examine data types
- Check missing values
- Check duplicate records
- Generate descriptive statistics
- Explore categorical variables
- Create a data dictionary
- Summarize initial observations

---

## Expected Output

A complete understanding of the dataset and its suitability for building CareerLens AI.

```python
import pandas as pd
import numpy as np

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", 1200)
```

```python
df = pd.read_csv("../data/raw/indian_tech_jobs_2026.csv")

df.head()
```

```python
rows, cols = df.shape

print("=" * 50)
print("DATASET OVERVIEW")
print("=" * 50)

print(f"Rows    : {rows:,}")
print(f"Columns : {cols}")
```

```python
columns = pd.DataFrame({
    "Column Number": range(1, len(df.columns) + 1),
    "Column Name": df.columns
})

columns
```

```python
df.info
```

```python
dtype_summary = (
    df.dtypes
      .value_counts()
      .rename_axis("Data Type")
      .reset_index(name="Count")
)

dtype_summary
```

```python
object_columns = df.select_dtypes(include="object").columns.tolist()
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

print("Categorical / Text Columns:")
print(object_columns)
print("Number of columns= ",len(object_columns))

print("\n" + "=" * 60 + "\n")

print("Numerical Columns:")
print(numeric_columns)
print("Number of columns= ",len(numeric_columns))
```

```python
missing_values = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": round((df.isnull().sum() / len(df)) * 100, 2)
})

missing_values = missing_values.sort_values(
    by="Missing Values",
    ascending=False
)

missing_values
```

```python
missing_only = missing_values[
    missing_values["Missing Values"] > 0
]

missing_only
```

```python
duplicates = df.duplicated().sum()

print("=" * 50)
print("DUPLICATE RECORDS")
print("=" * 50)

print(f"Duplicate Rows: {duplicates}")
```

```python
print("=" * 50)
print("MISSING VALUE SUMMARY")
print("=" * 50)

print(f"Columns with Missing Values : {(missing_values['Missing Values'] > 0).sum()}")
print(f"Total Missing Values        : {missing_values['Missing Values'].sum()}")
```

```python
numerical_summary = df.describe().T
numerical_summary
```

```python
categorical_summary = pd.DataFrame({
    "Unique Values": df.select_dtypes(include="object").nunique(),
    "Most Frequent": df.select_dtypes(include="object").mode().iloc[0]
})

categorical_summary
```
